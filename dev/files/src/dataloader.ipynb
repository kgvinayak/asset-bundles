```python
# Create widgets to get the parameters passed from the job.
# The second argument provides a default value.
dbutils.widgets.text("catalog"," ","Select catalog")
dbutils.widgets.text("schema"," ","Select schema")
dbutils.widgets.text("tabl"," ","Select table")
```

```python
  # Retrieve the parameters using dbutils.widgets
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
tabl = dbutils.widgets.get("tabl")
```

```python
print(f"Loading from {catalog}.{schema}.{tabl}")
```

```python
%sql

USE CATALOG IDENTIFIER(:catalog);
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:schema|| '.' ||:tabl)
(
   deptcode   INT,
   deptname  STRING,
   location  STRING
);
```

```python
%sql
INSERT INTO IDENTIFIER(:schema|| '.' ||:tabl) VALUES
   (10, 'FINANCE', 'EDINBURGH'),
   (20, 'SOFTWARE', 'PADDINGTON');
```
